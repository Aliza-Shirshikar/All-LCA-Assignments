```python
## import numpy as np

def create_matrix(rows, cols, matrix_num):
    print(f"Enter the elements for Matrix {matrix_num} ({rows}x{cols}):")
    elements = []
    for i in range(rows):
        row = []
        for j in range(cols):
            val = float(input(f"Element [{i+1}][{j+1}]: "))
            row.append(val)
        elements.append(row)
    return np.array(elements)

def main():
    print("--- Matrix Addition Program ---")
    try:
        
        rows = int(input("Enter the number of rows: "))
        cols = int(input("Enter the number of columns: "))
        
        if rows <= 0 or cols <= 0:
            print("Matrix dimensions must be positive integers.")
            return

       
        matrix_A = create_matrix(rows, cols, 1)
        matrix_B = create_matrix(rows, cols, 2)
        
        result_matrix = matrix_A + matrix_B
        
        print("--- Matrix 1 ---")
        print(matrix_A)
        
        print("--- Matrix 2 ---")
        print(matrix_B)
        
        print("--- Resultant Matrix (Sum) ---")
        print(result_matrix)
        
    except ValueError:
        print("Invalid input! Please enter integers for dimensions and numbers for elements.")

main()
```

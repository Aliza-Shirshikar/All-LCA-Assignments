```python
"""      Assignment 3           
Write a python program that accepts the length of three sides of a triangle as inputs. The
program should indicate whether or not the triangle is a right-angled triangle using function"""
```

```python
def check(side1, side2, side3):
    sides = sorted([side1, side2, side3])
    if (sides[0]**2 + sides[1]**2) == sides[2]**2:
        return True
    else:
        return False

def right_angled_triangle():
    print("--- Right-Angled Triangle Checker ---")
    try:
        a = float(input("Enter the length of the first side: "))
        b = float(input("Enter the length of the second side: "))
        c = float(input("Enter the length of the third side: "))
        if a <= 0 or b <= 0 or c <= 0:
            print("Side lengths must be greater than zero.")
            return
        if check(a, b, c):
            print("The triangle is a right-angled triangle.")
        else:
            print("The triangle is NOT a right-angled triangle.")
            
    except ValueError:
        print("Invalid input! Please enter numbers only.")
right_angled_triangle()
```

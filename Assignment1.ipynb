```python
"""                    ASSIGNMENT 1
Problem Statement: Student Data Management Using Python Collections
Write a Python program to create and store student information using a Dictionary, Tuple, and List. Perform the following operations on the Dictionary:
Add a new student record.
Delete an existing student record.
Update the details of a student.
Display the final student records after performing all the operations.
Use suitable student attributes such as Roll Number, Name, Branch, and Marks.

"""
```

```python
student = {}
def create(student):
    n = int(input("Enter the no. of student records to be added : "))
    for i in range(n):
            roll_no = len(student)+1
            name = input("Enter name of student : ")
            branch = input("Enter branch : ")
            marks = int(input("Enter marks : "))
            t = (name, branch, marks)
            student[roll_no]=t
    print(student)
def delete(student):
    d = int(input("Enter the roll no to be deleted : "))
    if d in student:
        del_student = student.pop(d)
        print(f"Deleted record for roll no. {d}({del_student[0]})")
    else:
        print(f"Roll no. {d} not found")
def add(student):
    roll_no = len(student)+1
    name = input("Enter name of student : ")
    branch = input("Enter branch : ")
    marks = int(input("Enter marks : "))
    t = (name, branch, marks)
    student[roll_no]=t
    print("Added")
def display(student):
    n=int(input("Enter 1 to view all data ,2 to view selected data : "))
    if n==1:
        print(student)
    elif n==2:
        no=int(input("Enter roll no of student : "))
        if no in student:
             print(f"Record for roll no. {no}({student[no]})")
        else:
             print(f"Roll no. {no} not found")
while True:
    n=int(input(""" 
------ Student Data Management ------
Enter :
1 - create 
2 - add
3 - delete
4 - display 
5 - exit 

Choice : """))
    if n == 1:
         create(student)
    elif n == 2:
         add(student)
    elif n == 3:
         delete(student)
    elif n == 4:
         display(student)
    else:
         break
             
         
```
